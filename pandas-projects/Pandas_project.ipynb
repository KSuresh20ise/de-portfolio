```python
import pandas as pd
import numpy as np
```

```python
raw = pd.DataFrame({
    "emp_id":     ["E001","E002","E003","E004","E005","E006"],
    "name":       ["Suresh","Raj",None,"Priya","Ankit","Sneha"],
    "department": ["Operations","IT","HR","IT","Operations",None],
    "salary":     [48000, 62000, 45000, None, 71000, 38000],
    "status":     ["Active","Active","Inactive","?","Active","Active"]
})
raw
```

```python
# Replacing ? in status with NaN
raw["status"] = raw["status"].replace("?", np.nan)
raw
```

```python
# Droping rows where name is missing
raw = raw.dropna(subset =["name"])
raw
```

```python
# Fill missing salary with department mean
raw["salary"] = raw.groupby("department")["salary"].transform(lambda x: x.fillna(x.mean()))
raw
```

```python
# Fill missing department with unknown
raw["department"] = raw["department"].fillna("Unknown")
raw
```

```python
# fill missing salary with overall mean (E006 has no department)
raw["salary"] = raw["salary"].fillna(raw["salary"].mean())
raw
```

```python
# Fill missing status with mode
raw["status"] = raw["status"].fillna(raw["status"].mode()[0])
print(raw)
```

```python
# Add hike_salary — 10% hike for Active, 0% for Inactive

raw["hike_salary"] = np.where(
    raw["status"] == "Active",
    raw["salary"] * 1.1,   # 10% hike if Active
    raw["salary"] * 1.0    # 0% hike if Inactive
)
raw
```

```python
# Add annual_bonus — 5% of salary for IT dept, 3% for others
raw["annual_bonus"] = np.where(
    raw["department"] == "IT",
    raw["salary"]*0.05,
    raw["salary"]*0.03)
raw
```

```python
# GroupBy department — show avg salary and count

summary = raw.groupby("department")["salary"].agg(["mean","count"])
summary
```

```python
# save cleaned data to CSV
raw.to_csv("employee_report.csv", index=False)
print("CSV Saved!")
```

```python
# Saving summary to JSON
import json

summary_dict = summary.to_dict()
with open("summary_report.json","w") as f:
    json.dump(summary_dict, f, indent=4)
print("JSON Saved")
```
