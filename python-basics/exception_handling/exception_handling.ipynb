Exception Handling allows a program to handle unexpected errors during execution in a controlled way, instead of crashing abruptly. It enables programs to detect errors, manage them properly and continue execution wherever possible.

```python
# try block contains code that may raise an exception
try:
    print(10/0)

# except block handles the exception raised by the try block
except ZeroDivisionError as e:
    print(f"Error: {e}")
    
# finally block executes whether an exception occurs or not.
# Useful for cleanup operations
finally:
    print("program execution completed!")
```

Built-in Exception Types - ZeroDivisionError, FileNotFoundError, ValueError, IndexError, KeyError, TypeError, AttributeError, ImportError, Exception

raise - Throwing Exception:
'raise' lets you throw an exception manually - used to validate data early and stop bad data from entering

```python
def driving_license_eligibility(age):
    if age <18:
        raise ValueError(f"Not eligible for license, age should be equal or above 18")
    return age

try:
    driving_license_eligibility(12)
except ValueError as e:
    print(f"Error: {e}")
```
