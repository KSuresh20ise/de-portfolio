```python
# lambda function anonymous function which doesnt require name and variable to stote
# syntax lambda arguments: expression
# No return statement is required
```

```python
# custom average function

def average(*x):
    return sum(x)/len(x)
print(average(1,2,3,4,5))
```

```python
# using lambda function
print((lambda x : sum(x)/len(x)) ([1,2,3,4,5]))
```

```python
avg = lambda x: sum(x)/len(x)
print(avg([1,2,3,4,5]))
```

```python
# Multiple parameters

power  =lambda x,y: x**y
print(power(2,3))
```

```python
# lambda function with iterables
# map() function applies to all elements in an iterable

alpha = ["a","b","c","d"]
captalize = map(lambda x: x.capitalize(), alpha)
print(list(captalize))
```
