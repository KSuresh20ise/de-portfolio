## Duplicates

Suppose the user entered the first , last name,address, height and weight later he entered again to change his height so duplicate rows formed to handle this we use duplicated() method

```python
import pandas as pd

data = {
    "first_name": ["Justin", "Justin", "Sarah", "Michael", "Sarah", "David"],
    "last_name": ["Saddlemyer", "Saddlemyer", "Connor", "Smith", "Connor", "Johnson"],
    "address": [
        "Boulevard du Jardin Botanique 3, Bruxelles",
        "Boulevard du Jardin Botanique 3, Bruxelles",
        "Baker Street 221B, London",
        "Wall Street 50, New York",
        "Baker Street 221B, London",
        "Park Avenue 12, Chicago"
    ],
    "height": ["193 cm", "194 cm", "168 cm", "175 cm", "168 cm", "182 cm"],
    "weight": ["87 kg", "87 kg", "55 kg", "70 kg", "55 kg", "78 kg"]
}

df = pd.DataFrame(data)
df
```

```python
columns = ['first_name','last_name','address']
duplicates = df.duplicated(subset=columns, keep=False)
```

```python
df[duplicates]
```

```python
# Droping duplicates only
df.drop_duplicates(inplace=True)
df
```
