```python
import pandas as pd
import datetime as dt
```

```python
rev = {
    "revenue":["20$","50$","150$","100$"]
}
df = pd.DataFrame(rev)
```

```python
df.sum()
```

1. Problem:  Above revenue in the string formate with $. so if we sum we wont get correct total revenue.
2. Solution: we need to remove the $ symbol and convert into integer

```python
df['revenue'] = df['revenue'].str.strip('$')
df
```

```python
# Converting into integer type using astype() method
df['revenue'] = df['revenue'].astype('int')
assert df['revenue'].dtype == ('int')
df.dtypes
```

```python
df.describe().round(2)
```

## Numerical or Categorical

```python
# 0. Never married, 1. married, 2. separated, 3. divorced
marriage = {
    'marriage_status':[2,1,3,0,1,2,2]
}
df = pd.DataFrame(marriage)
df
```

```python
# Above data is categorical if we do describe it shows the sum,mean,max but its worng no use of stats because its categorial data
df.describe()
```

```python
# So we need to convert into catogry type to get accurate and good quality results
df['marriage_status'] = df["marriage_status"].astype('category')
df.describe()
```
