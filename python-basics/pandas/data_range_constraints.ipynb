## Data constraints and ranges

Problem: There is movie rating data range between 1-5 if its above 6 then the data analysis cannot be accurate
Solution: If the rating is above 6 then convert into 5

```python
data = {
    "movie": [
        "The Shawshank Redemption",
        "The Dark Knight",
        "Inception",
        "Interstellar",
        "Pulp Fiction",
        "Fight Club",
        "Forrest Gump",
        "The Matrix",
        "Goodfellas",
        "Parasite",
        "Gladiator",
        "The Lord of the Rings",
        "Titanic",
        "Avatar",
        "The Avengers",
        "Whiplash",
        "The Prestige",
        "The Departed",
        "The Lion King",
        "Gladiator II"
    ],
    "rating": [8, 8, 7, 7, 6, 6, 5, 5, 4, 3, 7, 8, 4, 3, 6, 5, 7, 6, 5, 4]
}
df = pd.DataFrame(data)
df.head(5)
```

```python
df.dtypes
```

```python
# Converting the rating above 5 to 5

df.loc[df["rating"] > 5, "rating"] = 5
df
```

## Dates

```python
users = {
    "user_name":["Raj","John","Kishore","Paul"],
    "subscription_date":["6/12/2023","1/15/2021","5/5/2028","5/25/2027"]
}
df = pd.DataFrame(users)
df.head()
```

```python
df.dtypes
```

```python
# Convert subscription_date to date formate
df['subscription_date'] = pd.to_datetime(df['subscription_date']).dt.date
df.head()
```

```python
# convert subscription_date to todays date
today = dt.date.today()
df.loc[df['subscription_date'] > today, "subscription_date"] = today
df
```
