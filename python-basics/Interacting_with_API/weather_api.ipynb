API'S -> Application Program interface: An API is set of defined rules and protocals that allows different applications to communicate and exchange data with each other

In python we need requests package to make call to API'S: Using "pip install requests" command to install package

```python
import requests
import os
from dotenv import load_dotenv

load_dotenv(override=True)
```

1. So in order to make api request we need url and endpoint if url requireds API KEY as well
2. URL - http://api.weatherstack.com/
3. ENDPOINT - /current
4. it required api key as well always better to check documentation from the website which api your using in watherstack website - access_key = YOUR_API_KEY
5. query = CITY
6. Finally url looks like - http://api.weatherstack.com/current?access_key=YOUR_ACCESS_KEY&query=New York

```python
API_KEY = os.getenv("API_KEY")
```

```python
def weather(city):
    try:
        url = f"http://api.weatherstack.com/current?access_key={API_KEY}&query={city}"
        response = requests.get(url, timeout=10)
        response.raise_for_status()

        data = response.json()

        if data.get("success") is False:
            print("API Error:", data["error"]["info"])
            return None

        return data

    except requests.exceptions.RequestException as e:
        print("Request Error:", e)

    except Exception as e:
        print("Unexpected Error:", e)
```

```python
data = weather("Bengaluru")
```

```python
print("Location: ",data["location"]["name"])
print("Temperature:",data["current"]["temperature"])
print("Weather Description:",data["current"]["weather_descriptions"])
```
