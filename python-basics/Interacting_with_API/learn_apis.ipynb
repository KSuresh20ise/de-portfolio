```python
import requests
```

**Headers** -> Metadata sent along with an HTTP request/response. They describe the request
rather than carry the data itself. Examples:
- `Accept` / `Content-Type` -> the data format (e.g. application/json)
- `Accept-Language` -> preferred language (e.g. en-US)
- `Date` -> when the message was sent
- `Authorization` -> credentials/token

# URL - Universal Resource Locator

1. Protocol - https://
2. Domain - jsonplaceholder.typicode.com/
3. path - posts/
4. query params - 1

```python
# Adding headers to a request
response = requests.get(
    "https://jsonplaceholder.typicode.com/posts/11",
    headers={'accept':'application/json'}
)
```

```python
response.json()
```

```python
# Posting data

data = {'userId': 11,
  'id': 101,
  'title': 'Hello folks',
  'body': 'I hope you guys are doing well\nDont give up'
  }

post = requests.post(
    "https://jsonplaceholder.typicode.com/posts", json=data
)
print(post)
```

```python
# Adding headers to a request
headers = {'accept':'application/xml'}
response = requests.get(
    "https://jsonplaceholder.typicode.com/posts/4",
    headers=headers
)
```

```python
response.json()
```

```python
# Reading response headers
print(response.headers['content-type'])

# Acessing status code
print(response.status_code)
```

```python
# Check header what type of data accepts
print(response.headers.get('accept'))
```

# API AUTHENTICATIONS

1. Basic Authentication -> The client sends username:password, base64-encoded, in the Authorization header on every request. Base64 is encoding, not encryption, so HTTPS is mandatory.

```python
resp = requests.get(
    "https://api.example.com/v1/policies",
    auth=("suresh", "my_password"),   # requests builds the header for you
    timeout=10,
)
# Sent as: Authorization: Basic c3VyZXNoOm15X3Bhc3N3b3Jk
```

2. API Key/Token Auth: A static secret string issued to an application. The client sends it with every request, and the server looks it up to identify the caller.

```python
resp = requests.get(
    "https://api.example.com/v1/policies",
    headers={"X-API-Key": API_KEY},
    timeout=10,
)
```

3. Bearer Token / JWT: The client logs in once and receives a token, then sends Authorization: Bearer <token> on later requests. A JWT (JSON Web Token) is a signed token with three parts, header.payload.signature. The payload carries claims such as user ID and expiry, so the server can verify it without a database lookup.

```python
# Step 1: log in, get token
login = requests.post(
    "https://api.example.com/auth/login",
    json={"username": "suresh", "password": "my_password"},
    timeout=10,
)
token = login.json()["access_token"]

# Step 2: use token
resp = requests.get(
    "https://api.example.com/v1/policies",
    headers={"Authorization": f"Bearer {token}"},
    timeout=10,
)
```

4. OAuth 2.0
Definition: A delegation framework. A user (or a service) authorizes an app to access resources with limited scopes, without sharing a password. It issues short-lived access tokens and long-lived refresh tokens.

```python
token_resp = requests.post(
    "https://auth.example.com/oauth/token",
    data={
        "grant_type": "client_credentials",
        "client_id": "my_client_id",
        "client_secret": "my_client_secret",
        "scope": "policies.read",
    },
    timeout=10,
)
access_token = token_resp.json()["access_token"]

resp = requests.get(
    "https://api.example.com/v1/policies",
    headers={"Authorization": f"Bearer {access_token}"},
    timeout=10,
)
```
