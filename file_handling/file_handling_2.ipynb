```python
import csv

# write csv

with open("policies.csv","w") as f:
    writer = csv.writer(f)
    
    # write header
    writer.writerow(["policy_id","holder","type","premium","status"])

    #write rows
    writer.writerow(["POL001","Suresh","Health",540,"Active"])
    writer.writerow(["POL002","Raj","Auto",1540,"Active"])
    writer.writerow(["POL003","Priya","Cyber",5440,"Inactive"])
    writer.writerow(["POL004","Ramesh","Health",1040,"Active"])
    
```

```python
# Read CSV
with open("policies.csv","r") as f:
    reader = csv.reader(f)

    header = next(reader)
    print(f"Columns: {header}")
    for row in reader:
        print(row)
```

```python
# DictReader -> rows as dictionaries - Best
with open("policies.csv","r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["holder"], row["premium"])
```

```python
# DictWriter → write from dictionaries
policies = [
    {"policy_id": "POL001", "holder": "Suresh",
     "type": "Health", "premium": 4500},
    {"policy_id": "POL002", "holder": "Raj",
     "type": "Auto", "premium": 2300}
]
with open("policies.csv","w", newline = "") as f:
    fields = ["policy_id","holder","type","premium"]
    writer = csv.DictWriter(f, fieldnames = fields)
    writer.writeheader()
    writer.writerows(policies)
```
