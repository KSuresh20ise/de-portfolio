```python
def average_v(values):
    try:
        avg = sum(values)/len(values)
        return avg
    except:
        print("average(value): It accepts list or set as arguments. Please provide set or list")
print(average_v([10,20,30]))
```

```python
# Raise TypeError

def average(values):
    if type(values) in (set, list):
        return sum(values)/len(values)
    else:
        raise ValueError("average() accepts list or set. Please provide list or set")
```

```python
print(average(10))
```
