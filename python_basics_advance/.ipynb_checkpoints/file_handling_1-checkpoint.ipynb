```python
# File Handling - Basic reading
filename = "C:/Users/kvsur/Downloads/data_retrivel.txt"
file  = open(filename, mode="r")
print(file.read())
file.close()
```

```python
# Using with we no need to close manually its automatically closes the file

with open("C:/Users/kvsur/Downloads/data_retrivel.txt", "r") as file:
    print(file.read())
```

```python
# readline() method reads single line from the text document
with open("C:/Users/kvsur/Downloads/data_retrivel.txt", "r") as file:
    for line in file:
        print(file.readline())
    
```

```python
# writing file
with open("policies.txt","w") as f:
    f.write("POL001 Suresh Health 4500\n")
    f.write("PLOO2 Raj Auto 2300\n")

# write multiple lines

lines = [
    "POL003 Siva Cyber 600\n",
    "POL004 Mahi Health 5200"
]

with open("policies.txt","a") as f:
    f.writelines(lines)

with open("policies.txt","r") as f:
    print(f.read())
```

```python
# Loop line by line ✅ BEST for large files!
with open("policies.txt", "r") as f:
    for line in f:
        print(line.strip())  # strip removes \n
```
